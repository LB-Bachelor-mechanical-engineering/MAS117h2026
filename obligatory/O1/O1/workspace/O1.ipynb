### Oppgave 2.1
En universitetscampus har 20 klasserom og 200 kontorer. I hvert klasserom finnes 18 lysrør som bruker $110\text{ W}$ elektrisk effekt per lysrør. Kontorene har to lysrør hver. Universitetet er åpent 240 dager i året. Klasserommene og kontorene brukes ikke i seks timer per døgn, men lyset står på hele døgnet. Anta at strømprisen er $1{,}5\text{ NOK/kWh}$.

Hvor mye sparer universitetet pr. år hvis alt lyset slås av i de seks timene verken klasserommene eller kontorene brukes?

```python
dEds_lysroer = 118 #[w]

N_klasserom=20
N_Lysrør_d_klasserom=18
dEds_klasserom:float = N_klasserom * N_Lysrør_d_klasserom * dEds_lysroer

N_kontorer=200
N_Lysrør_d_kontorer=2
dEds_kontorer:float = N_kontorer * N_Lysrør_d_kontorer * dEds_lysroer

dEds:float = dEds_kontorer + dEds_klasserom

kwh:float = dEds * 1e-3

time_saved_d_day = 6 #h

days = 240

tot_kwh_d_year = days * time_saved_d_day * kwh

pris_d_kwh = 1.5

spart = pris_d_kwh * tot_kwh_d_year

print(f"dersom du skrur av lyset {time_saved_d_day} timer hver dag")
print(f"så sprer du {spart:0.2f} kroner på et år")
```

### Oppgave 2.2
Vann pumpes fra en innsjø til en stor lagertank som på toppen er åpen til atmosfæren. Vannoverflaten i tanken ligger $20\text{ m}$ over vannoverflaten av innsjøen. Volumstrømmen er $60\text{ liter/s}$ og det brukes $17{,}5\text{ kW}$ elektrisk effekt for det. Se bort fra tap i strømningsenergi på grunn av friksjon og fra endringer i vannets kinetiske energi. Bruk $\rho = 997\text{ kg/m}^3$ som tetthet for flytende vann.

* **a)** Beregn den totale virkningsgraden til pumpe-motor-enheten.

```python
# F * dx = E
# F = m * a, a=g=9.81
# m = Rho * V
Rho = 997 #[kg/m³]
dVdt= 60 *1e-3 #[m³/s]
g = 9.81 #[m/s²]
dx = 20 #m
# dEdt = E/t
# = F * dx / t
# = m * g * dx / t
# = Rho * V * g * dx / t   
# = Rho * V / t * g * dx   
# dEdt = Rho * V / t * g * dx

dEdt = Rho * dVdt * g * dx

motor_dEdt=17.5*1e3 #[W]

ratio = dEdt / motor_dEdt

print(f"Effekt faktor er {ratio:0.5f}")
```

* **b)** Beregn trykkdifferansen mellom innløp og utløp av pumpen som er nødvendig for å pumpe vannet opp til tanken.

```python
p= 20 * 997 * 9.81 #Pa
print(f"trykket på en vannsøyle på 20 meter blir {p:0.2f}")
```

### Oppgave 2.3
En elektrisk motor med en maksimal effekt på $75\text{ hk}$ (hestekrefter, engelsk: hp, horse power) på akslingen har en virkningsgrad på $91\%$. Den skal erstattes med en ny motor, som har samme maksimale akslingseffekt, men en effektivitet på $95{,}4\%$. Den gamle motoren er i drift i $4368\text{ timer}$ per år ved $75\%$ av dens maksimale effekt (altså "75% last" eller "75% lastfaktor"). Den nye motoren skal gå med samme last og samme antall timer.


* **a)** Hvor mye elektrisk energi spares per år ved å erstatte den gamle med den nye motoren?

```python
E_motor1 = 75 / 0.91 * 746 * 4368 * 60**2 * 0.75 

# anntar relativ mekanisk last fra
# motor 2 sitt perspektiv gir mindre enn 75% av maks last

# E1*h1*
h_motor2 = E_motor1 * 0.91 / (75 * 746 * 4368 * 60**2)


E_motor2 = 75 / 0.954 * 746 * 4368 * 60**2 * h_motor2

E = E_motor1 - E_motor2 #ws

print(h_motor2)
print(f"du sparer {E/1e9:0.2f} [Giga][J] på et år med å bytte til motor 2")
```

```python
# MAS117 - Oppgave 2.3
# Beregning av energisparing og kostnadsbesparelse ved motorbytte

# Gitte verdier
P_maks_hk = 75.0          # Maksimal akslingseffekt i hestekrefter (hp)
eta_gammel = 0.91         # Virkningsgrad for gammel motor
eta_ny = 0.954            # Virkningsgrad for ny motor
driftstimer = 4368        # Driftstimer per år (h/år)
lastfaktor = 0.75         # Lastfaktor (75 % last)
strompris = 1.5           # Strømpris i NOK/kWh

# Omregningsfaktor: 1 hp = 0.7457 kW
HP_TIL_KW = 0.746

# 1. Akslingseffekt ved 75 % last (kW)
P_aksling_hp = P_maks_hk * lastfaktor
P_aksling_kW = P_aksling_hp * HP_TIL_KW

# 2. Tilført elektrisk effekt for begge motorene (kW)
# P_el = P_aksling / virkningsgrad
P_el_gammel = P_aksling_kW / eta_gammel
P_el_ny = P_aksling_kW / eta_ny

# a) Elektrisk energi spart per år (kWh)
# Energi = Effekt * tid
E_gammel = P_el_gammel * driftstimer
E_ny = P_el_ny * driftstimer
E_spart = E_gammel - E_ny

# b) Utgiftsreduksjon per år (NOK)
kostnad_spart = E_spart * strompris

# Utskrift av resultater
print("--- Resultater for Oppgave 2.3 ---")
print(f"Maksimal akslingseffekt: {P_aksling_kW:.2f} kW ({P_aksling_hp:.2f} hp ved 75% last)")
print(f"Elektrisk effekt gammel motor: {P_el_gammel:.3f} kW")
print(f"Elektrisk effekt ny motor:    {P_el_ny:.3f} kW")
print("-" * 35)
print(f"a) Innspart elektrisk energi per år: {E_spart:.2f} kWh")
print(f"b) Innspart beløp per år:            {kostnad_spart:.2f} NOK")
```
