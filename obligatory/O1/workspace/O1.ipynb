### Oppgave 1.1
Et manometer, som er festet til en tank, viser et vakuumtrykk på $70\text{ kPa}$, mens det atmosfæriske trykk måles med et barometer til å være $1020\text{ mbar}$.

Hva er absolutttrykket i tanken i:
- **a)** $\text{kPa}$
- **b)** $\text{N/m}^2$
- **c)** $\text{lbf/in}^2$
- **d)** $\text{mmHg}$

Vis hvilke omregningsfaktorer du bruker! Noter hvor du har tatt omregningsfaktorene fra eller vis utledningen dersom du har beregnet de selv.

```python
def PascalFromBar(bar):
    #Anntar at 1 Bar er 10e5 Pa
    return bar * 1e5

def BarFromPascal(pascal):
    return pascal / PascalFromBar(1)

def PsiFromPascal(pascal):
    return pascal * 6894.7572932

def PascalFromPsi(psi):
    return psi / PsiFromPascal(1)

def MmHgFromPascal(pascal, g=9.81, Rho_Hg=13600):
    return pascal / (g * Rho_Hg)

def PascalFromMmHg(mmHg, g=9.81, Rho_Hg=13600):
    return mmHg / MmHgFromPascal(1)

P_tank = 70e3 #Pa
P_atmosphere = PascalFromBar(1020e-3) #Pa

P = P_tank + P_atmosphere

print(f"a) absolutt trykk i Kpa er: {P*1e-3} [KPa]")
print(f"b) absolutt trykk i N/m² er: {P} [N/m²]")
print(f"c) absolutt trykk i lbs/in² er: {PsiFromPascal(P)} [lbs/in²]")
print(f"d) absolutt trykk i mmHg er: {MmHgFromPascal(P)} [mmHg]")
```

### Oppgave 1.2

*(Illustrasjon: Bygg med høyde $h$, $675\text{ mmHg}$ på toppen og $695\text{ mmHg}$ ved bunnen)*

- **a)** Beregn høyden av bygget i $\text{m}$.

```python
Rho_Hg=13600 #kg/m³
Rho_Air=1.18 #kg/m³
dh=995e-3 - 975e-3 #m

print(f"Høyden på bygningen totalt er:{(Rho_Hg*dh/Rho_Air):.2f}[m]")
```

- **b)** I noen tabeller er massetettheten til kvikksølv angitt med $13\,550\text{ kg/m}^3$. Hvor stor blir forskjellen mellom resultatet i a) og beregningen med denne verdien i henholdsvis $\text{m}$ og prosent ($\%$)?

```python
Rho_Hg1=13600 #kg/m³
Rho_Hg2=13550 #kg/m³
Rho_Air=1.18 #kg/m³
dh_Hg: float=995e-3 - 975e-3 #m

dh_Air1: float=Rho_Hg1*dh/Rho_Air
dh_Air2: float=Rho_Hg2*dh/Rho_Air

Ratio: float=dh_Air2/dh_Air1
print(f"Høyden på bygningen var først:{(dh_Air1):.2f}[m]")
print(f"Høyden på bygningen med ny tabellverdi er:{(dh_Air2):.2f}[m]")
print(f"relativ forskjel i prosent er ca.:{((1-Ratio)*1e2):.2f}[%]")
```


### Oppgave 1.3
Den indre og ytre overflate av en vindusrute av vanlig glass, som er $4\text{ mm}$ tykk og har et areal på $2\text{ m} \times 2\text{ m}$, har en temperatur på henholdsvis $22^\circ\text{C}$ og $8^\circ\text{C}$. Varmeledningskoeffisienter finner du i tabell 2-3 i læreboken.

- **a)** Beregn energimengden som tapes som varme gjennom vindusruten i løpet av $12\text{ timer}$.

```python

dt:float = 12 * 60 * 60 #s

thickness:float= 4e-3 #m

height:float = 2 #m
width:float = 2 #m
area:float = height * width

dT:float = (22+273) - (8+273) #K
kT:float = 1.05 #W/(mK)

loss_W: float = kT*dT*area/thickness
loss_E: float = loss_W*dt #w*s -> J

print(f"gjennomsnittlig effekttap av varme energi er {loss_W:0.2f} [W]")
print(f"totalt tapt energi er {loss_E:0.2f} [J]")
```

```python

dt:float = 12 * 60 * 60 #s

thickness:float= 4e-3 #m

height:float = 2 #m
width:float = 2 #m
area:float = height * width

dT:float = (22+273) - (8+273) #K
kT:float = 1.05 #W/(mK)

loss_W: float = kT*dT*area/thickness
loss_E: float = loss_W*dt #w*s -> J

print(f"gjennomsnittlig effekttap av varme energi er {loss_W:0.2f} [W]")
print(f"totalt tapt energi er {loss_E:0.2f} [J]")
```
