# Reversible steady flow

Work done during a process depends o nthe path followed, and the properties at the end states.

$$W_{b, out}= \int^2_1{P}dV $$

a positive result indicates work done by the system on the outside  
*(a positive amount of energy is moving outwards)*.

negative result indicates work done on the system by the outside  
*(a negatie amount of energy is moving outwards)*.

taking energy inputs in the form of both heat and work as positive quantities, the energy balance for a steady-floW device undergoing
an internally reversible process can be expressed in the differential form as:

$$ \delta q_{rev} + \delta w_{rev} = dh + dke + dpe $$

- $\delta q_{rev}$ is the reversible thermal energy added to the system (energy in)

- $\delta q_{rev}$ is the reversible mechanical energy added to the system (energy in)

- $dh$ is the increase of enthalpy of the medium currently in the system.

    $h = U + PV$
    - $U$ is internal energy (chemical, nuclear, etc.)
    - $P$ is pressure
    - $V$ is volume

- $dke$ is the increase of kinetic energy of the medium currently in the system.

    $ke=\frac{v^2}{2}=\int{v}dv $
    - $v$ is the velocity

- $dpe$ is the increase of potential energy of the medium currently in the system.

    $pe=\int{g}dy$
    - $g$ is the acceleration from gravity
